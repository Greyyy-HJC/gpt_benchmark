```python
import numpy as np
import gvar as gv
from lametlat.utils.plot_settings import *
from lametlat.utils.resampling import *
from lametlat.preprocess.read_raw import pt2_to_meff
```

```python
# corr_conf_ls = gv.load("dump/quark_prop_conf_ls_gfix_hyp.dat")
corr_conf_ls_fhf = gv.load("dump/quark_prop_conf_ls_gfix_hyp_gfix.dat")
# corr_conf_ls = gv.load("dump/quark_prop_conf_ls_no_cg.dat")
print(np.shape(corr_conf_ls_fhf))

corr_jk = jackknife(corr_conf_ls_fhf)
corr_gv = jk_ls_avg(corr_jk)

meff_fhf = pt2_to_meff(corr_gv, boundary="none")

fig, ax = default_plot()
ax.errorbar(np.arange(len(meff_fhf)), gv.mean(meff_fhf), gv.sdev(meff_fhf), label="meff", **errorb)
ax.set_xlabel(r"$t$", **fs_p)
ax.set_ylabel(r"$m_{\mathrm{eff}}$", **fs_p)
ax.legend(**fs_small_p)
# ax.set_ylim(-2, 2.2)
plt.tight_layout()
plt.show()
```
